# FashionStyle14 Dataset2 - Complete CSV

Scan all images under `FashionStyle14_v1/dataset2` and write `complete_dataset2.csv`.


## 1. Setup and Imports


```python
import os
import pandas as pd
from pathlib import Path

pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)

print("Libraries imported successfully!")
print(f"Current working directory: {os.getcwd()}")
```

```python
main_path = Path("../FashionStyle14_v1")
dataset_path = main_path / "dataset2"
output_csv = main_path / "complete_dataset2.csv"

if not dataset_path.is_dir():
    raise FileNotFoundError(f"Dataset folder not found: {dataset_path.resolve()}")
```

## 2. Dataset Discovery - Create Complete CSV


```python
def discover_all_images(dataset_path):
    """Discover all images in the dataset folder and create a complete dataframe."""
    print(f"Scanning dataset folder: {dataset_path}")

    all_images = []
    supported_formats = ('.jpg', '.jpeg', '.png', '.JPG', '.JPEG', '.PNG')

    style_folders = [
        f for f in os.listdir(dataset_path)
        if os.path.isdir(os.path.join(dataset_path, f))
    ]

    print(f"Found {len(style_folders)} style categories: {sorted(style_folders)}")

    for style_folder in sorted(style_folders):
        style_path = os.path.join(dataset_path, style_folder)
        image_count = 0

        for image_file in os.listdir(style_path):
            if image_file.startswith('._'):
                continue
            if image_file.lower().endswith(supported_formats):
                image_path = os.path.join(style_path, image_file)
                all_images.append({
                    'image_path': image_path,
                    'style': style_folder,
                    'filename': image_file,
                })
                image_count += 1

        print(f"  {style_folder}: {image_count} images")

    df = pd.DataFrame(all_images)
    print(f"\nTotal images discovered: {len(df)}")
    return df


complete_dataset2 = discover_all_images(dataset_path)
```

```python
print("Complete Dataset Information:")
print(f"Shape: {complete_dataset2.shape}")
print(f"Columns: {list(complete_dataset2.columns)}")
print("\nFirst 5 rows:")
print(complete_dataset2.head())

print("\nStyle distribution:")
style_counts = complete_dataset2['style'].value_counts().sort_index()
print(style_counts)

complete_dataset2['image_path'] = (
    complete_dataset2['image_path']
    .astype(str)
    .str.replace('FashionStyle14_v1\\', '', regex=False)
    .str.replace('FashionStyle14_v1/', '', regex=False)
)

complete_dataset2[['image_path']].to_csv(output_csv, index=False, header=False)
print(f"\nComplete dataset saved to '{output_csv}'")
print(f"Total images: {len(complete_dataset2)}")
```
